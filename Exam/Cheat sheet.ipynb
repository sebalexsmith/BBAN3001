**Explaination:**
Linear regression models the relationship between a dependent and one or more independent variables. The goal is to find the coefficients which minimize the differences between the predicted and actual values.

### Naive Bayes Classification

**Uses:**
Is used for assigning instances (strings, numbers, documents) to one of multiple predefined classes. Text classification, spam filtering, customer segmentation, fraud detection are some common uses for naive bayes classification.

**Explaination:**
Naive Bayes classification works by calculating the probability of an event or class based on initial conditions and other features using Bayes' Theorem:

$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$, where A is what you want to predict and B is the set of observed features.

Ex: classifying spam emails, classes are either spam or not spam, the deciding features are the words in the email. The probability is calculated based on the frequency of certain words and consequently classified to the most likely class: spam or not spam.

**Approach:**

### K-Means Clustering

**Uses:**
Grouping something based on several features, customer segmentation, grouping based on purchasing behavior, demographics etc. Anomaly detection, identifying outliers or unusual patterns in data. Identifying groups of products frequently bought together. Identifying groups of stocks with similar price movements.

**Explaination:**
Number of clusters (K) is chosen, K random cluster centerpoints are chosen. The distances from each datapoint to each of the K cluster centerpoints is calculated, and each datapoint is assigned to the cluster whose centerpoint is the closest. The cluster centerpoints are recalculated based on the datapoints assigned to it. By reclassifying each datapoint to each cluster and reclaculating the centerpoints several times, the centerpoints will converge on the centers of each cluster. If a datapoint is closer to one centerpoint than another, it belongs to that cluster, and all datapoints are thereby classified.

**Approach:**

```python
from sklearn.cluster import KMeans
```
