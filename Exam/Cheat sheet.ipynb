```python
# Other useful imports for analyzing
from sklearn.metrics import accuracy_score # How well predictions fared against actual. Evaluating models.
from sklearn.metrics import confusion_matrix # For evaluating the accuracy of a classification.
from sklearn.metrics import r2_score # Regression score function, coefficient of determination.
from sklearn.preprocessing import LabelEncoder # For encoding labels to integer classes, dummy+.
```

```python
# Pandas create dataframe:
df = pd.DataFrame(data = 'Data as 2D array or dictionary')
```

```python
# Import data from different data files (csv and excel):
df = pd.read_csv('filename.csv', delimiter = ',', decimal = '.')
df = pd.read_excel('filename.xlsx', sheet_name = 'excel_sheet_name', decimal = '.')
```

```python
# Show dataset:
df, df.head(), df.tail(), df.columns, df.shape, df.size
```

```python
# Describe dataset:
df.info() # For general content of the dataset, datatypes, number of columns and instances of each feature.
df.describe() # For mean, std, min and max of each feature in the dataset.
```

```python
# Convert object features to dummy variables.
df = pd.get_dummies(df, columns = ['list of columns to be converted'], drop_first = True, dtype = int)
```

```python
# Label Encoder use (import like above):
encoder = LabelEncoder()

df['Column 1 classes'] = encoder.fit_transform(df['Column 1'])
```

```python
# Train-Test Split:
from sklearn.model_selection import train_test_split
y = df['Dependent Variable']
X = df.drop(['Dependent Variable'], axis = 1) # or
X = df[df.drop('Dependent Variable', axis = 1).columns]

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.8, random_state = 42, shuffle = True)
# Check that everything looks good by checking each new dataframe X_train, X_test, y_train, y_test.
```

### Logistic Regression

**Uses:**
Commonly used for binary classification problems, whether an oservation belongs to one of two classes. Used to estimate the probability of an observation belonging to a particular class, assessing the likelyhood of an event happening.

**Explaination:**
Modeling the relationship between the independent variables and the probability of belonging to a particular class. Output is a probability score, then a threshold separates the score to make the classification.

```python
from sklearn.linear_model import LogisticRegression
```
