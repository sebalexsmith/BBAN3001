### Explaination of General Statistical Properties

- Mean: The average, the most common value in a set of data.
- Std: A measure of the amount of variation in a set of data. Low means that most of the data is close close to the mean.
- Kurtosis: How flat the distribution of data is. Negative means the distribution is flatter.
- Skewness: How symmetric a distribution of data is. Positive skew means the larger amount of data is to the left of the mean.
- R-squared: The coefficient of determination, the higher the better a feature explains a phenomenon. Using weight to explain size, if R-squared is high that means weight explains size well.
- Overfitting: Overfitting occurs when a ML model supposed to describe a phenomenon only works well on the training data, but when it tries to predict other data to describe the same phenomenon, it does not work well. It has high variance.
- Underfitting: When a model is to simple to explain the complexities in data, high bias.
- Bias: Is the inability of a ML model to capture the true relationship. Minimize the bias.
- Variance: Difference of sums of squares for different datasets describing the same phenomenon. Minimize the variance.
- ROC and AUC: ROC (Reciever Operator Characteristic), AUC (Area Under the Curve).

**Explaination:**
Linear regression models the relationship between a dependent and one or more independent variables. Naive Bayes classification works by calculating the probability of an event or class based on initial conditions and other features using Bayes' Theorem:

Ex: classifying spam emails, classes are either spam or not spam, the deciding features are the words in the email. The probability is calculated based on the frequency of certain words and consequently classified to the most likely class: spam or not spam.

### K-Means Clustering

**Uses:**
Grouping something based on several features, customer segmentation, grouping based on purchasing behavior, demographics etc. Anomaly detection, identifying outliers or unusual patterns in data. Identifying groups of products frequently bought together. Identifying groups of stocks with similar price movements.

**Explaination:**
Number of clusters (K) is chosen, K random cluster centerpoints are chosen. The distances from each datapoint to each of the K cluster centerpoints is calculated, and each datapoint is assigned to the cluster whose centerpoint is the closest. The cluster centerpoints are recalculated based on the datapoints assigned to it. By reclassifying each datapoint to each cluster and reclaculating the centerpoints several times, the centerpoints will converge on the centers of each cluster. If a datapoint is closer to one centerpoint than another, it belongs to that cluster, and all datapoints are thereby classified.

**Approach:**

```python
from sklearn.cluster import KMeans
```

Import the data and make any necassary changes. Scale the data as done in the data handling section.

Scaling is important when using K-Means clustering because the algorithm is based on distance calculations, and when some features are larger than others, it can disproportionatly influence the result. Scaling the data ensures that all features in the data contributes equally to the final result.

Plot the data to visualize the possible clusters. 

```python
# Find the optimal value for K
K_based_inertia = []
K_range = range(1, 10)

for k in K_range:
    model = KMeans(n_clusters = k, n_init = 10, random_state = 42)
    model = model.fit(df[['Scaled data 1', 'Scaled data 2...']])
    K_based_inertia.append(model.inertia_)
```

```python
# Plot the inertia and find the elbow point for the optimal value of K:
plt.plot(K_range, K_based_inertia, marker = 'o')
plt.xlabel('K value')
plt.ylabel('Score based on K')
plt.show()
```

```python
# Define and fit the model with the optimal K-value:
model = KMeans(n_clusters = 'Optimal K', n_init = 10, random_state = 42)
model = model.fit(df[['Scaled data 1', 'Scaled data 2...']])
# n_init is the amount of times the algorithm is run with different initial centerpoints, the best result is used, for larger, higher dimetional data use larger n_init. random_state is so that the algorithm picks the same starting centerpoints whenever it is run.
```

```python
# Add prediction of cluster to the dataset:
df['Cluster'] = model.predict(df[['Scaled data 1', 'Scaled data 2...']])
```

```python
# Show the estimated coordinates of the cluser centerpoints:
centers = model.cluster_centers_
pd.DataFrame(centers).rename(columns = {0: 'x', 1: 'y...'})
```

```python
# Plot the clusters for 2d data:
dfn = df[df['Cluster'] == 'n']

plt.scatter(dfn['Scaled data 1'], dfn['Scaled data 2'], color = 'Red')

plt.scatter(centers[:, 0], centers[:, 1], color = 'black', marker = 'x')

plt.xlabel('Scaled data 1')
plt.ylabel('Scaled data 2')
plt.show()
```

### Decision Tree Classification

**Uses:**
Binary classification, multiple classification, classifying features even when the data is imbalanced as in there are a lot more instances of one feature than another. Identifying the most influential features in the decision-making process. Seeing non-linear patterns in data.  

**Explaination:**
A decision tree which classifies things into categories. Recursively parting the data into subsets based on the most consequential features. From the tree root data is further parted by branches and eventually comes to the leaves, or the correct classification. The features for the deciding nodes are chosen based on criteria like gini impurity to separate instances into distinct classes. 

**Approach:**

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_val_score
```

```python
# Define and fit the model (maybe change parameters to make the model better? min_samples_split):
model = DecisionTreeClassifier().fit(X_train, y_train)
```

```python
# Evaluate the model:
model.score(X_train, y_train)
model.score(X_test, y_test)
```

```python
# Make confusion matrix to show the anomalies in predictions for both training and test sets:
y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

matrix = confusion_matrix(y_pred_train, y_train, normalize = 'all')
plt.figure(figsize = ('x', 'y'))
heatmap(matrix, annot = True, fmt = '.2%', cmap = 'Blues')
plt.xlabel('Predictions')
plt.ylabel('Actual')
plt.show()

matrix = confusion_matrix(y_pred_test, y_test, normalize = 'all')
plt.figure(figsize = ('x', 'y'))
heatmap(matrix, annot = True, fmt = '.2%', cmap = 'Blues')
plt.xlabel('Predictions')
plt.ylabel('Actual')
plt.show()
```

```python
# Find importances, what features are the most consequential for the model, and plot in bar chart:
importances = pd.DataFrame(model.feature_importances_, columns = ['Importances'], index = X_train.columns)

importances.sort_values(by = ['Importances'], axis = 0, ascending = False, inplace = True)

importances.plot(kind = 'bar')
plt.show()

# Comment on the importances.
```

```python
# Plot preliminary decision tree:
fig = plt.figure(figsize = ('x', 'y'))
plot_tree(model, feature_names = X_train.columns, filled = True, fontsize = 10)
plt.show()
```

```python
# Start pruning the tree (minimal cost complexity pruning). Create the model from scratch:
model = DecisionTreeClassifier()
```

```python
# Define path:
path = model.cost_complexity_pruning_path(X_train, y_train)

ccp_alphas = path.ccp_alphas[:] # Remove any unnecessary alpha values (?).
print(ccp_alphas) # Make sure everything looks good.
```

```python
# Make a list of models with each having a different alpha from the ccp_alphas list:
models = []

for alpha in ccp_alphas:
    model = DecisionTreeClassifier(ccp_alpha = alpha).fit(X_train, y_train)
    models.append(model)
```

```python
# Plot the models scores to compare alpha values:
train_scores = [model.score(X_train, y_train) for model in models]
test_scores = [model.score(X_test, y_test) for model in models]

plt.plot(ccp_alphas, train_scores, marker = '.', label = 'Train', color = 'blue')
plt.plot(ccp_alphas, test_scores, marker = '.', label = 'Test', color = 'red')
plt.xlabel('Alpha')
plt.ylabel('Score')
plt.legend()
plt.show()

# But this is only for one configuration of training and test sets, so the alpha at the intersection point might not be the best it can be.
```

```python
# Cross validation to find the optimal value for alpha:
models_cv = []

for alpha in ccp_alphas:
    model = DecisionTreeClassifier(ccp_alpha = alpha)
    scores = cross_val_score(model, X_train, y_train, cv = 5) # Five-fold cross validation
    models_cv.append([alpha, np.mean(scores), np.std(scores)])
```

```python
# Create dataframe with results from the cross validation:
cv_results = pd.DataFrame(models_cv, columns = ['Alpha', 'Mean Accuracy', 'STD Accuracy'])
cv_results.head()
```

```python
# Plot the results from the cross validation:
plt.plot(cv_results['Alpha'], cv_results['Mean Accuracy'], marker = '.', color = 'red')
plt.xlabel('Alpha')
plt.ylabel('Mean Accuracy')
plt.show()
```

```python
# Find optimal alpha:
optimal_alpha = cv_results[cv_results['Mean Accuracy'] == max(cv_results['Mean Accuracy'])]['Alpha']
optimal_alpha
```

```python
# Define optimal alpha:
optimal_alpha = float(optimal_alpha['Row if there are multiple alphas'])
```

```python
# Defining and fitting the pruned decision tree:
pruned_model = DecisionTreeClassifier(ccp_alpha = optimal_alpha).fit(X_train, y_train)
```

```python
# Plot the pruned decision tree:
plt.figure(figsize = ('x', 'y'))
plot_tree(pruned_model, feature_names = X_train.columns, filled = True, fontsize = 10)
plt.show()
```

```python
# Score the pruned model:
pruned_model.score(X_train, y_train)
pruned_model.score(X_test, y_test)
```

```python
# Make predictions with pruned model and a confusion matrix for train and test sets:
pruned_pred_train = pruned_model.predict(X_train)
pruned_pred_test = pruned_model.predict(X_test)

matrix = confusion_matrix(pruned_pred_train, y_train, normalize = 'all')
plt.figure(figsize = ('x', 'y'))
heatmap(matrix, annot = True, fmt = '.2%', cmap = 'Blues')
plt.xlabel('Predictions')
plt.ylabel('Actual')
plt.show()

matrix = confusion_matrix(pruned_pred_test, y_test, normalize = 'all')
plt.figure(figsize = ('x', 'y'))
heatmap(matrix, annot = True, fmt = '.2%', cmap = 'Blues')
plt.xlabel('Predictions')
plt.ylabel('Actual')
plt.show()
```

**Explaination:**
Artificial neural networks consist of interconnected nodes which are organized into layers (input, hidden, and output). Each node in each layer is connected with weights which represent the strength of the connection. Each node also has a bias which can affect the sum of the input to the node to make the network more adaptable to complex situations. In training these connection weights and biases are adjusted to optimize the performance of the network. In addition to the weights and biases, each node may use and activation function, these functions determine the output of the node based on the input, some examples of activation functions are sigmoid, ReLU, and linear.
